# 7. Reto 4 — `print()` vs `return`

### Objetivo

Comparar dos funciones que realizan tareas diferentes.

Crear:

````python
def mostrar_resultado(valor):
    print(valor)
````

y:

````python
def obtener_resultado(valor):
    return valor
````

### Actividad

Utiliza ambas funciones y explica:

> ¿Cuál permite que otro bloque del programa utilice el resultado?

### Evidencia

Incluye el código y una explicación breve.

````python
dato_1 = mostrar_resultado(7)    # imprime 7
dato_2 = obtener_resultado(7)   # devuelve 7

print("Lo que guardó mostrar_resultado:", dato_1)
print("Lo que guardó obtener_resultado:", dato_2)
print("Usando el resultado en otro cálculo:", dato_2 + 3)
````

# 8. Reto 5 — Función de clasificación

### Contexto

Dataset de Trabajo ya utiliza clasificación de valores.

### Objetivo

Crear o revisar una función:

````python
LIMITE_ALTO = 100  
LIMITE_BAJO = 10  


def clasificar_valor(valor):
    valor = int(valor)
    if valor > LIMITE_ALTO:
        clasificacion = "ALTO"
    elif valor < LIMITE_BAJO:
        clasificacion = "BAJO"
    else:
        clasificacion = "NORMAL"
    return clasificacion
````

Debe conservar las reglas utilizadas actualmente en Dataset de Trabajo.

### Pruebas

Prueba valores que produzcan:


```text
ALTO
NORMAL
BAJO
```


### Requisito

No duplicar la lógica de clasificación fuera de la función.

### Pregunta

¿Por qué es mejor reutilizar `clasificar_valor()`?

````python
print(clasificar_valor(424)) 
print(clasificar_valor(55))   
print(clasificar_valor(2))    
````

### Respuesta

Reglas actuales usadas: **ALTO** si los casos son mayores a 100, **BAJO** si son menores a 10 y **NORMAL** en los demás casos.

Es mejor reutilizar `clasificar_valor()` porque la regla queda escrita en un solo lugar. Si el límite cambia (por ejemplo de 100 a 150), se modifica una sola vez y todo el programa se actualiza. Copiar la lógica en varios sitios hace más probable que unos queden con la regla vieja y otros con la nueva, y vuelve el código más largo y difícil de leer.

# 15. Reto 12 — Crear `procesar_registros()`

### Contexto

La lógica que procesa registros no debería estar mezclada con la lógica
del menú.

### Objetivo

Separar el procesamiento en:

### Funciones de apoyo

Para que `procesar_registros()` funcione se conservan las validaciones de la Semana 6 y se agregan los datos del dataset.

````python
def es_numero_valido(numero):
    if str(numero).isdigit():
        return True
    else:
        return False


def esta_en_rango(valor, minimo, maximo):
    if not str(valor).isdigit():
        return False
    valor = int(valor)
    if valor < minimo or valor > maximo:
        return False
    else:
        return True


def es_codigo_valido(codigo):
    codigo = str(codigo).strip()
    if " - " not in codigo:
        return False
    prefijo = codigo.split(" - ")[0]
    if len(prefijo) != 2 or not prefijo.isdigit():
        return False
    return esta_en_rango(prefijo, 1, 21)


def es_rango_valido(texto):
    texto = str(texto).strip().lower()
    if texto.startswith("mayor de"):
        numero = texto[8:].strip()
        return esta_en_rango(numero, 0, 120)
    partes = texto.split(" a ")
    if len(partes) != 2:
        return False
    inicio = partes[0].strip()
    fin = partes[1].strip()
    if not esta_en_rango(inicio, 0, 120) or not esta_en_rango(fin, 0, 120):
        return False
    if int(inicio) > int(fin):
        return False
    else:
        return True


def es_registro_valido(registro):
    if not es_codigo_valido(registro["LOCALIDAD_RESIDENCIA"]):
        return False
    if not es_rango_valido(registro["GRUPOS_EDAD"]):
        return False
    if registro["SEXO"] != "M" and registro["SEXO"] != "F":
        return False
    if not esta_en_rango(registro["ESTRATO_SOCIOECONOMICO"], 1, 6):
        return False
    if not es_numero_valido(registro["CASOS"]):
        return False
    if not es_numero_valido(registro["MUERTES"]):
        return False
    return True
````

````python
def cargar_registros():
    registros = [
        {"LOCALIDAD_RESIDENCIA": "09 - Fontibón", "GRUPOS_EDAD": "45 A 49",
         "SEXO": "M", "ESTRATO_SOCIOECONOMICO": 2, "CASOS": 2, "MUERTES": 0},
        {"LOCALIDAD_RESIDENCIA": "15 - Antonio Nariño", "GRUPOS_EDAD": "40 a 44",
         "SEXO": "F", "ESTRATO_SOCIOECONOMICO": 6, "CASOS": 1, "MUERTES": 0},
        {"LOCALIDAD_RESIDENCIA": "04 - San Cristóbal", "GRUPOS_EDAD": "55 a 59",
         "SEXO": "M", "ESTRATO_SOCIOECONOMICO": 2, "CASOS": 424, "MUERTES": 16},
        {"LOCALIDAD_RESIDENCIA": "11 - Suba", "GRUPOS_EDAD": "20 a 24",
         "SEXO": "M", "ESTRATO_SOCIOECONOMICO": "Sin Dato", "CASOS": 443, "MUERTES": 0},
        {"LOCALIDAD_RESIDENCIA": "Sin Dato", "GRUPOS_EDAD": "80 a 84",
         "SEXO": "F", "ESTRATO_SOCIOECONOMICO": 2, "CASOS": 2, "MUERTES": 0},
        {"LOCALIDAD_RESIDENCIA": "06 - Tunjuelito", "GRUPOS_EDAD": "Mayor de 100",
         "SEXO": "F", "ESTRATO_SOCIOECONOMICO": 3, "CASOS": 2, "MUERTES": 0},
        {"LOCALIDAD_RESIDENCIA": "10 - Engativá", "GRUPOS_EDAD": "10 a 14",
         "SEXO": "X", "ESTRATO_SOCIOECONOMICO": 5, "CASOS": 3, "MUERTES": 0},
        {"LOCALIDAD_RESIDENCIA": "08 - Kennedy", "GRUPOS_EDAD": "5 a 9",
         "SEXO": "F", "ESTRATO_SOCIOECONOMICO": 3, "CASOS": -4, "MUERTES": 0},
    ]
    return registros


def obtener_columna(registros, campo):
    valores = []
    for registro in registros:
        valores.append(int(registro[campo]))
    return valores
````

### Función `procesar_registros()`

````python
def procesar_registros(registros):
    validos = []
    invalidos = []
    suma_casos = 0     
    existe_alto = False     

    for registro in registros:
        if es_registro_valido(registro):
            validos.append(registro)
            suma_casos += int(registro["CASOS"])
            if clasificar_valor(registro["CASOS"]) == "ALTO":
                existe_alto = True
        else:
            invalidos.append(registro)

    casos_validos = obtener_columna(validos, "CASOS")
    altos, normales, bajos = contar_por_clasificacion(casos_validos)

    return validos, invalidos, altos, normales, bajos, suma_casos, existe_alto
````

# 16. Reto 13 — Función de resumen

### Objetivo

Separar la presentación del resumen.

Crear una función que reciba los resultados y los muestre.

Ejemplo:

````python
def mostrar_resumen(total, validos, invalidos, porcentaje, hombres, mujeres,
                    suma_casos, suma_muertes, tasa, hay_muertes):
    print("===== RESUMEN =====")
    print("Total de registros:", total)
    print("Registros válidos:", validos)
    print("Registros inválidos:", invalidos)
    print("Porcentaje de registros inválidos:", porcentaje, "%")
    print("Hombres (M):", hombres)
    print("Mujeres (F):", mujeres)
    print("Suma de casos (válidos):", suma_casos)
    print("Suma de muertes (válidos):", suma_muertes)
    print("Tasa de mortalidad:", round(tasa, 2), "%")
    print("¿Existe algún registro válido con muertes?", hay_muertes)
    print("¿Existe algún registro inválido?", invalidos > 0)
````

### Requisito

La función debe concentrarse en mostrar la información.

### Pregunta

¿Por qué es conveniente separar el cálculo del resumen de su presentación?

````python
def contar_por_sexo(validos, sexo):
    contador = 0
    for registro in validos:
        if registro["SEXO"] == sexo:
            contador += 1
    return contador


def existe_muerte(validos):
    bandera = False
    for registro in validos:
        if int(registro["MUERTES"]) > 0:
            bandera = True
    return bandera


def calcular_porcentaje(parte, total):
    if total > 0:
        return parte / total * 100
    else:
        return 0
````

````python
def mostrar_procesamiento(registros):
    validos, invalidos, altos, normales, bajos, suma_casos, existe_alto = procesar_registros(registros)
    posicion = 0
    for registro in registros:
        posicion += 1
        if es_registro_valido(registro):
            print("Registro", posicion, "-> Válido")
        else:
            print("Registro", posicion, "-> Inválido")
    print("Total de registros:", len(registros))
    print("Registros válidos:", len(validos))
    print("Registros inválidos:", len(invalidos))


def resumir(registros):
    validos, invalidos, altos, normales, bajos, suma_casos, existe_alto = procesar_registros(registros)
    total = len(registros)
    porcentaje = calcular_porcentaje(len(invalidos), total)
    hombres = contar_por_sexo(validos, "M")
    mujeres = contar_por_sexo(validos, "F")
    suma_muertes = calcular_suma(obtener_columna(validos, "MUERTES"))
    tasa = calcular_porcentaje(suma_muertes, suma_casos)
    hay_muertes = existe_muerte(validos)
    mostrar_resumen(total, len(validos), len(invalidos), porcentaje, hombres,
                    mujeres, suma_casos, suma_muertes, tasa, hay_muertes)


def mostrar_clasificacion(altos, normales, bajos):
    print("===== CLASIFICACIÓN DE CASOS =====")
    print("ALTO:", altos)
    print("NORMAL:", normales)
    print("BAJO:", bajos)


def clasificar_todos(registros):
    validos, invalidos, altos, normales, bajos, suma_casos, existe_alto = procesar_registros(registros)
    mostrar_clasificacion(altos, normales, bajos)
````

### Respuesta

Conviene separar el cálculo de la presentación porque:

- El cálculo se puede **reutilizar** (mostrarlo en pantalla, guardarlo en un archivo o usarlo en otra función) sin cambiar nada.
- Se puede cambiar **cómo se ve** el resumen sin riesgo de dañar los cálculos.
- Las funciones de cálculo retornan valores, así que son fáciles de **probar** comparando el resultado esperado con el obtenido.

# 26. Preguntas de reflexión

Responde con tus propias palabras:

1. ¿Qué problema resuelven las funciones?
2. ¿Cuál es la diferencia entre un parámetro y un argumento?
3. ¿Por qué `return` es importante?
4. ¿Cuándo conviene utilizar una función?
5. ¿Qué diferencia existe entre `print()` y `return`?
6. ¿Qué ventaja tiene dividir Dataset de Trabajo en funciones?
7. ¿Por qué no es conveniente depender innecesariamente de variables
   globales?
8. ¿Qué significa refactorizar?
9. ¿Qué riesgo existe al refactorizar un programa que ya funciona?
10. ¿Por qué fue útil crear un commit antes de comenzar la refactorización?

### Respuestas

1. **¿Qué problema resuelven las funciones?** Evitan repetir código, le dan nombre a una tarea y dividen un problema grande en partes pequeñas que se pueden entender y probar por separado.
2. **Parámetro y argumento:** el parámetro es la variable que aparece en la definición (`def saludar(nombre)`); el argumento es el valor real que se envía al llamar la función (`saludar("Ana")`).
3. **¿Por qué `return` es importante?** Porque entrega el resultado a quien llamó la función, para guardarlo o usarlo en otro cálculo. Sin `return` la función devuelve `None`.
4. **¿Cuándo conviene una función?** Cuando una tarea se repite o cuando una parte del programa tiene una responsabilidad clara que se puede nombrar, reutilizar y probar.
5. **`print()` vs `return`:** `print()` solo muestra el valor en pantalla; `return` lo devuelve para que el programa lo siga usando.
6. **Ventaja de dividir el Dataset en funciones:** el código queda más ordenado y con menos duplicación, es más fácil de probar y corregir, y `main()` queda corto y fácil de explicar.
7. **¿Por qué no depender de variables globales?** Porque cualquier parte del programa puede cambiarlas, lo que dificulta encontrar errores y hace que las funciones dependan de algo que no se ve. Con parámetros y `return` cada función es independiente y reutilizable.
8. **¿Qué significa refactorizar?** Reorganizar el código por dentro (funciones, nombres, estructura) para mejorarlo sin cambiar lo que hace.
9. **Riesgo al refactorizar:** introducir errores o cambiar el comportamiento sin darse cuenta en un programa que ya funcionaba. Por eso se prueban los resultados antes y después.
10. **¿Por qué fue útil el commit antes de empezar?** Porque deja guardada la versión que funcionaba: si algo sale mal se puede comparar o volver a ella, y sirve como punto de referencia para verificar que el comportamiento no cambió.
